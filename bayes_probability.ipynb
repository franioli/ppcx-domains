# Bayes formula:
    
    P(A|B)  = P(B|A) * P(A) / P(B) = 

where: 
- P(A|B) = P(Collapse|PCA_Component): probability of having a Collpase given the observed PCA principal component value (posterior).
- P(B|A) = P(PCA_Component|Collapse): probability of observing the PCA principal component value given that a Collapse occurs (likelihood).
- P(A) = P(Collapse): prior probability of a Collapse occurring (prior).
- P(B) = P(PCA_Component). overall probability of observing the PCA principal component value (evidence)

1. P(A) — Prior probability of collapse
P(A) is estimated from the historical data as the proportion of days with a collapse (P(Collapse) = n_collapse / n_total_days).

2. P(B|A) — Likelihood: density of PCA index given collapse

Since the PCA index is continuous, this is a probability density function (PDF), not a simple ratio. You estimate it from the ~200 PCA index values observed on collapse days. Two common approaches:

```python 
from scipy.stats import gaussian_kde

# pca_at_collapse: array of PCA index values on the ~200 collapse days
kde_collapse = gaussian_kde(pca_at_collapse)
# For a given day's PCA value x:
p_B_given_A = kde_collapse(x)
```

3. P(B) — Evidence: overall density of PCA index

Same approach, but using all days (both collapse and non-collapse):
```python 

# pca_all: array of PCA index values for ALL days
kde_all = gaussian_kde(pca_all)
p_B = kde_all(x)
```

4. Posterior: P(A|B)
```python 
p_A = n_collapse_days / n_total_days
posterior = kde_collapse(x) * p_A / kde_all(x)
```
