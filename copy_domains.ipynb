```python
%load_ext autoreload
%autoreload 2

import re
import shutil
from pathlib import Path

output_dir = Path("output")
dir_pattern = r"PPCX_Tele_\d{4}-\d{2}-\d{2}$"
domains_geometry_file_pattern = "*sectors_final.geojson"

domains_dir = Path("domains")
domains_dir.mkdir(exist_ok=True)
```

```python
year_dirs = sorted(output_dir.iterdir())
print(f"Found {len(list(year_dirs))} year directories.")
```

```python
for year_dir in year_dirs:
    if not year_dir.is_dir():
        continue

    print(f"Processing year directory: {year_dir}")
    dest_dir = domains_dir / year_dir.name
    dest_dir.mkdir(exist_ok=True)

    for day_dir in year_dir.iterdir():
        if not day_dir.is_dir():
            continue
        if not re.match(dir_pattern, day_dir.name):
            continue

        # Find the geometry file
        geometry_files = list(day_dir.glob(domains_geometry_file_pattern))
        if not geometry_files:
            print(f"No geometry file found in {day_dir}")
            continue
        geometry_file = geometry_files[0]

        # Copy to destination
        shutil.copy(geometry_file, dest_dir)
```
